## Manipulação dos dados

```python
df = df.drop(columns=['Unnamed: 0','name', 'category','goal','pledged','usd pledged'])
df.head()
```

```python
df.dtypes
```

```python
#df['usd_goal_real']= df['usd_goal_real'].str.replace("USD ", "")
df['usd_goal_real'].apply(lambda x: x.strip("USD ")) #Caso não haja uma função existente, aqui se cria a função para a troca
```

```python
df['usd_goal_real'] = df['usd_goal_real'].apply(lambda x: x.strip("USD ")) #Aqui é a aplicação no dataframe
df['usd_goal_real'] = df['usd_goal_real'].astype('int64') #Mudando de string para inteiro
```

```python
df.head()
```

```python
df['launched'].str.replace(" \d\d:\d\d","",regex=True)#Usando RegEx \d representa um numero
```

```python
df['launched']= df['launched'].str.replace(" \d\d:\d\d","",regex=True)
```

```python
df['launched']= pd.to_datetime(df['launched'], format='%d/%m/%y')
df['deadline']= pd.to_datetime(df['deadline'], format='%d/%m/%y')
```

```python
df.head()
```

```python
df.dtypes
```

```python
(df["deadline"]-df["launched"]).dt.days #dt.days apenas retira a palavra days depois do numero
```

```python
df["time_range"] = (df["deadline"]-df["launched"]).dt.days
df.head()
```
