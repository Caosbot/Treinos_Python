## Com Bonston.csv

```python
df_Boston = pd.read_csv(r"F:\Cursos\DNC\Banco de dados\Boston.csv", sep = ",")
```

```python
df_Boston.head()
```

```python
df_Boston.drop(['Unnamed: 0'], axis=1, inplace=True)
```

```python
target= df_Boston['medv']
target
```

```python
df_Boston.drop(columns=['medv'], inplace=True)
```

```python
df_Boston.head()
```

```python
X = df_Boston.values
y = target.values
```

### Modelo



1.   Regreção Linear do SKLEARN
2.   Suport Vector Regression SKLEARN
3.   Decision Tree Regression XGBoost

Modeling Assumptions: Apenas variaveis numéricas
Sem Hiperparametros para os 3 modelos (Default)





```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from xgboost import XGBRegressor
```

```python
from sklearn.svm import SVR
```

#### Dataset Split 20%

```python
X_train, X_test, y_train, y_test  = train_test_split(X,y, train_size = 0.8, test_size=0.2,random_state = 42)
```

#### Treino

```python
regLinear = LinearRegression().fit(X_train,y_train) #Testando o modelo
yLinear =  regLinear.predict(X_test) #Para predizer os valores de casas
```

### Tecnica 1 . Regressão Linear

#### Métrica MSE e RMSE
Mean score error
Root Mean Squared Error
Para penalizar erros de previsão

```python
MSELinear = mean_squared_error(y_test, yLinear) #Comparando os valores
```

```python
print(f"MSE: {MSELinear}")
print(f"RMSE: {np.sqrt(MSELinear)}") #Erro em torno de até
```

### Tecnica 2. SVR

```python
regSVR = SVR().fit(X_train,y_train) #Treinando o modelo
ySVR = regSVR.predict(X_test) #Testando o modelo
```

```python
MSESVR = mean_squared_error(y_test, ySVR)
```

```python
print(f"MSE: {MSESVR}")
print(f"RMSE: {np.sqrt(MSESVR)}") #Erro em torno de até
```

### Tecnica 3 Decision Tree Regression (XGBosst)

```python
regXGB = XGBRegressor().fit(X_train,y_train) #Treinando o modelo
yXGB = regXGB.predict(X_test) #Testando o modelo
```

```python
MSEXGB =mean_squared_error(y_test, yXGB)
```

```python
print(f"MSE: {MSEXGB}")
print(f"RMSE: {np.sqrt(MSEXGB)}") #Erro em torno de até
```

## Otimização de Hiperparametros

Aplicando metodo de GridSearch do SKLEARN

```python
regXGB.get_params().keys()
```

```python
parametros = {
    "max_depth": [5,6,7], "learning_rate":[0.1, 0.2, 0.3,],
    "objective": ['reg:squarederror'], "booster": ['gbtree'],
    "n_jobs": [5], "gamma": [0,1], "min_child_weight":[1,3],
    "max_delta_step": [0,1], "subsample": [0.5, 1]
    }
```

```python
xgbGrid = GridSearchCV(XGBRegressor(), parametros, refit='neg_mean_square_error', verbose=True)
```

```python
xgbGridModel = xgbGrid.fit(X_train,y_train)
```

```python
xgbGridModel.best_params_
```

```python
yGrid = xgbGridModel.predict(X_test)
```

```python
MSEGrid = mean_squared_error(y_test,yGrid)
```

```python
print(f"MSE XBG: {MSEGrid}")
print(f"RMSE XGB: {np.sqrt(MSEGrid)}") #Erro em torno de até
```
